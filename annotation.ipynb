```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt

batter=pd.read_csv(r'C:\Users\Homes\Desktop\matplotlib\batter.csv')
sample_df = batter.head(100).sample(25, random_state=5)
sample_df
```

```python
x=[1,2,3,4,5]
y=[5,6,7,8,9]

plt.scatter(x,y)
plt.Text(1,5,'point=1')
plt.Text(2,6,'point=2')
plt.Text(3,7,'point=3')
```

```python
plt.figure(figsize=(18,10))

plt.scatter(
    sample_df['avg'],
    sample_df['strike_rate'],
    s=sample_df['runs']
)

for i in range(sample_df.shape[0]):
    plt.text(
        sample_df['avg'].values[i],
        sample_df['strike_rate'].values[i],
        sample_df['batter'].values[i]
    )

plt.show()
```

```python
# horizontal and vertical line
plt.figure(figsize=(18,10))

plt.scatter(
    sample_df['avg'],
    sample_df['strike_rate'],
    s=sample_df['runs']
)

for i in range(sample_df.shape[0]):
    plt.text(
        sample_df['avg'].values[i],
        sample_df['strike_rate'].values[i],
        sample_df['batter'].values[i]
    )

plt.axhline(130,color='red')
plt.axvline(30,color='red')
```
