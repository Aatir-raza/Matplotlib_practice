```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r'C:\Users\Homes\Desktop\matplotlib\batter.csv')

fig, ax = plt.subplots(nrows=1, ncols=2, sharex=True)

ax[0].scatter(df['avg'], df['strike_rate'], df['runs'])
ax[1].scatter(df['avg'],df['strike_rate'],df['runs'])
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2, sharex=True)

ax[0].scatter(df['avg'], df['strike_rate'], df['runs'])
ax[0].set_title('Avg vs strike rate')
ax[0].set_ylabel('strike rate')

ax[1].scatter(df['avg'], df['runs'])
ax[1].set_title('Avg vs runs')
ax[1].set_ylabel('runs')
ax[1].set_xlabel('avg')

plt.show()
```
