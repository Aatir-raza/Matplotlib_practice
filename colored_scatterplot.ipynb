```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt


iris = pd.read_csv(r'C:\Users\Homes\Desktop\matplotlib\iris.csv')

species_num = iris['Species'].astype('category').cat.codes

plt.scatter(
    iris['SepalLengthCm'],
    iris['PetalLengthCm'],
    c=species_num
)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.show()
```

```python
plt.figure(figsize=(15,7))
plt.scatter(
    iris['SepalLengthCm'],
    iris['PetalLengthCm'],
    c=species_num,
    s=100
)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.figure(figsize=(15,7))
plt.show()
```
